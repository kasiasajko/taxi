- trip_start_timestamp, trip_end_timestamp: strings instead of datatype
- trip_seconds: string instead of a number, but these look like decimal type
- pickup_census_tract, dropoff_census_tract: scientific number, but stored as string
- pickup_community_area: these seem like numbers, but are stored as string
- dropoff_community_area: sometimes these are string but sometimes numbers
- pickup_latitude, pickup_longitude: these should be decimal numbers but are strings
- pickup_location: consists of 'POINT ' and then a pair of numbers in brackets. example: POINT (-87,6251921424 41,8788655841)
- dropoff_latitude, dropoff_longitude: should be number
- dropoff_location: consists of 'POINT ' and then a pair of numbers in brackets. example: POINT (-87,6251921424 41,8788655841)
